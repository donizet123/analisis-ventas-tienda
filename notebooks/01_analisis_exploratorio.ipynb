```python
df_limpio["vendedor"] = df_limpio["vendedor"].fillna("No especificado")

df_limpio["metodo_pago"] = df_limpio["metodo_pago"].fillna("No especificado")
```

```python
df_limpio = df_limpio.dropna(subset=["cantidad"])
```

```python
df_limpio[df_limpio["precio_unitario"] < 0]
```

```python
df_limpio = df_limpio[df_limpio["precio_unitario"] >= 0]
```

```python
df_limpio["producto"] = df_limpio["producto"].str.strip().str.title()
```

```python
df_limpio["fecha"] = pd.to_datetime(df_limpio["fecha"])
```

```python
df_limpio["cantidad"] = pd.to_numeric(df_limpio["cantidad"])
```

```python
df_limpio["importe_venta"] = (
    df_limpio["cantidad"] * df_limpio["precio_unitario"]
)
```

```python
# Guarda tu archivo limpio manteniendo el formato original de Windows (Latin-1)
df_limpio.to_csv(
    "../data/ventas_tienda_dataset_limpio.csv", 
    index=False, 
    encoding="latin1",  # <--- USA LATIN1 AQUÍ TAMBIÉN
    sep=","
)

print("¡Archivo limpio guardado en el formato original compatible con tu MySQL!")
```

```python
# Resumen general de las ventas

total_ventas = df_limpio["importe_venta"].sum()
total_operaciones = len(df_limpio)
total_unidades = df_limpio["cantidad"].sum()
ticket_promedio = total_ventas / total_operaciones

print("RESUMEN GENERAL DE VENTAS")
print("-------------------------")
print(f"Importe total vendido: S/ {total_ventas:,.2f}")
print(f"Cantidad de operaciones: {total_operaciones}")
print(f"Unidades vendidas: {total_unidades}")
print(f"Importe promedio por operación: S/ {ticket_promedio:,.2f}")
```

```python
productos_vendidos = (
    df_limpio.groupby("producto")["cantidad"]
    .sum()
    .sort_values(ascending=False)
)

print(productos_vendidos.head(10))
```

```python
ingresos_producto = (
    df_limpio.groupby("producto")["importe_venta"]
    .sum()
    .sort_values(ascending=False)
)

print(ingresos_producto.head(10))
```

```python
ventas_categoria = (
    df_limpio.groupby("categoria")["importe_venta"]
    .sum()
    .sort_values(ascending=False)
)

print(ventas_categoria)
```

```python
ventas_vendedor = (
    df_limpio.groupby("vendedor")["importe_venta"]
    .sum()
    .sort_values(ascending=False)
)

print(ventas_vendedor)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))

sns.barplot(
    x=ventas_categoria.values,
    y=ventas_categoria.index
)

plt.title("Ingresos por categoría")
plt.xlabel("Importe de ventas (S/)")
plt.ylabel("Categoría")

plt.tight_layout()
plt.show()
```
